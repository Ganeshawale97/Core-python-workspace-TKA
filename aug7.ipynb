## Division
1. / (Normal Division) -> Division of two numbers result is always in float.
2. // (Floor Division) -> Floor Division Returns Integer Part of result.

/ (Normal Division) 

```python
a = 6
b = 2
res = a/b
print(a , type(a))
print(b , type(b))
print(res , type(res))
```

// (Floor Division)

```python
a = 5
b = 2
res = a//b
print(a , type(a))
print(b , type(b))
print(res , type(res))
```

## Modulus / Modulo / mod ---> %

```python
a = 123
b = 10
res = a%b
print(a , type(a))
print(b , type(b))
print(res , type(res))
```

### original num = 123
### after floor div by 10 = 12
### after mod by 10 = 3

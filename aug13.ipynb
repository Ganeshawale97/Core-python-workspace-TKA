# String in Python

```python
s1 = 'pune'
s2 = "India"
s3 = '''  #Menu
1. Idli
2. Dosa 
3. Vada
4. Pongal'''

print(s1)
print(s2)
print(s3)
print(type(s1))
print(type(s2))
print(type(s3))
print(s1[0])
print(len(s1))
```

# Slicing

```python
s = "Instagram"

print(len(s))
print(s[1])
print(s[-1])
print(s[3],s[4], s[5])
print(s[-6],s[-5], s[-4])

substring = s[3]+s[4]+s[5]
print(substring)


```

# Indexing

```python
s1 = "instagram"
s2 = "Facebook"

s3 = s1 + " " + s2
print(s3)
#print(s1 - s2)  {type error}
print(s1 * 3)

s3 = "ten"
num = 10
print(s3 == num)
print(s1[3:6])  
print(s1[3 : len(s1)])  
print(s2[4:7])

print('Note' + s2[4: len(s2)])
```

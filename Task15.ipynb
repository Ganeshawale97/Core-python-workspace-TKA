```python
num = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]

# 1. Print all elements from the list.
print(num)

#2. Print the first and last element.
print("first element:", num[0])
print("Last element:", num[-1])

# 3. Find the total number of elements without manually counting them.
print("Total number of element:", len(num))

# 4. Calculate the sum of all elements.
print("Sum of all Elements:", sum(num))

# 5. Find the largest and smallest number.
print("Maximum Number:", max(num))
print("Minimum Number:", min(num))

# 6. Print all even numbers.
even = []
for n in num:
    if n % 2 == 0:
        print("Even number:", n)
        even.append(n)

# 7. Print all odd numbers.
odd = []
for n in num:
    if n % 2 == 1:
        print("Odd number:", n)
        odd.append(n)

# 8. Count how many even and odd numbers are present.
print("Number of even numbers:", len(even))
print("Number of odd numbers:", len(odd))

# 9. Check whether 23 exists in the list.
print( "23 exists in the list:",23 in num)

# 10. Find the index position of 89.
print("Index position of 89:", num.index(89))

# 11. Add 100 at the end of the list.
num.append(100)
print("Updated list:", num)

# 12. Insert 50 at index position 3.
num.insert(3, 50)
print("Updated list:", num)

# 13. Remove the first occurrence of 45.
num.remove(45)
print("Updated list:", num)

# 14. Remove duplicate values from the list.
unique = list(set(num))
print(unique)

# 15. Sort the list in ascending and descending order.
num.sort()
print("Ascending order:", num)

num.sort(reverse=True)
print("Descending Order:", num)



```

```python
num = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]

# 16. Create a new list containing the square of every number.
squares = []
for n in num:
    squares.append(n ** 2)
print("Squares:", squares)

# 17. Create a new list containing only numbers greater than 30.
greater_30 = []
for n in num:
    if n > 30:
        greater_30.append(n)
print("Numbers greater than 30:", greater_30)

# 18. Print the list in reverse order without using reverse().
print("Reversed list:", num[::-1])

# 19. Find the second-largest number.
unique_sorted = sorted(set(num), reverse=True)
print("Second largest number:", unique_sorted[1])

# 20. Find the second-smallest number.
unique_sorted_asc = sorted(set(num))
print("Second smallest number:", unique_sorted_asc[1])

# 21. Count how many times each number appears.
counts = {}
for n in num:
    counts[n] = counts.get(n, 0) + 1
print("Number counts:", counts)

# 22. Separate the numbers into two lists: even numbers and odd numbers.
evens = []
odds = []
for n in num:
    if n % 2 == 0:
        evens.append(n)
    else:
        odds.append(n)
print("Even list:", evens)
print("Odd list:", odds)

# 23. Find all duplicate elements.
seen = set()
duplicates = set()
for n in num:
    if n in seen:
        duplicates.add(n)
    seen.add(n)
print("Duplicate elements:", list(duplicates))

# 24. Find all elements that appear only once.
once = []
for n in set(num):
    if num.count(n) == 1:
        once.append(n)
print("Elements appearing once:", once)

# 25. Replace every number smaller than 20 with 0.
replaced = []
for n in num:
    if n < 20:
        replaced.append(0)
    else:
        replaced.append(n)
print("List after replacement:", replaced)
```

```python
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]

# 26. Print all student names.
print(students)

# 27. Print names that start with "A".
starts_a = [name for name in students if name.startswith("A")]
print(starts_a)

# 28. Print names containing more than five characters.
long_names = [name for name in students if len(name) > 5]
print(long_names)

# 29. Convert every name to uppercase.
upper_names = [name.upper() for name in students]
print(upper_names)

# 30. Sort the names alphabetically.
print(sorted(students))

# 31. Find the longest student name.
longest = max(students, key=len)
print(longest)

# 32. Check whether "Sneha" exists in the list.
print("Sneha" in students)

# 33. Create a new list containing the length of every name.
lengths = [len(name) for name in students]
print(lengths)

# 34. Remove "Raj" from the list.
students.remove("Raj")
print(students)

# 35. Print all names in reverse alphabetical order.
print(sorted(students, reverse=True))
```

```python
# 36. Merge two lists without using the + operator.
list1 = [1, 2, 3]
list2 = [4, 5, 6]
merged = list1.copy()
for item in list2:
    merged.append(item)
print("Merged list:", merged)

# 37. Find common elements between two lists.
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
common = []
for n in list1:
    if n in list2:
        common.append(n)
print("Common elements:", common)

# 38. Find elements present in the first list but not in the second.
only_in_first = []
for n in list1:
    if n not in list2:
        only_in_first.append(n)
print("Elements only in first list:", only_in_first)

# 39. Rotate a list one position to the right.
lst = [1, 2, 3, 4, 5]
rotated = [lst[-1]] + lst[:-1]
print("List rotated by one position:", rotated)

# 40. Rotate a list by a user-given number of positions.
lst = [1, 2, 3, 4, 5]
k = int(input("Enter number of positions: "))
k = k % len(lst)
rotated_k = lst[-k:] + lst[:-k] if k != 0 else lst
print("Rotated list:", rotated_k)

# 41. Move all zero values to the end of a list.
lst = [0, 1, 0, 3, 12, 0, 5]
non_zero = []
zeros = []
for n in lst:
    if n != 0:
        non_zero.append(n)
    else:
        zeros.append(n)
result = non_zero + zeros
print("List with zeros moved to end:", result)

# 42. Divide a list into smaller lists of three elements each.
lst = [1, 2, 3, 4, 5, 6, 7, 8]
chunks = []
for i in range(0, len(lst), 3):
    chunks.append(lst[i:i+3])
print("List divided into chunks of 3:", chunks)

# 43. Check whether a list is a palindrome.
lst = [1, 2, 3, 2, 1]
is_palindrome = lst == lst[::-1]
print("Is the list a palindrome:", is_palindrome)

# 44. Find two numbers whose sum equals a user-given target.
lst = [2, 7, 11, 15]
target = int(input("Enter target sum: "))
found = False
for i in range(len(lst)):
    for j in range(i + 1, len(lst)):
        if lst[i] + lst[j] == target:
            print("Pair found:", lst[i], "and", lst[j])
            found = True
if not found:
    print("No pair found")

# 45. Remove duplicate values while maintaining the original order.
lst = [1, 3, 2, 3, 1, 4, 5, 4]
seen = set()
no_duplicates = []
for n in lst:
    if n not in seen:
        no_duplicates.append(n)
        seen.add(n)
print("List without duplicates (order preserved):", no_duplicates)
```

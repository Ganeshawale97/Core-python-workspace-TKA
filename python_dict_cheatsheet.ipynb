# Python Dictionary Cheatsheet
A quick reference for common `dict` operations in Python.

## Creating a dictionary

```python
person = {"name": "Alice", "age": 30, "city": "New York"}

# Alternative creation methods
person2 = dict(name="Bob", age=25, city="Boston")
empty = {}
from_keys = dict.fromkeys(["a", "b", "c"], 0)  # {'a': 0, 'b': 0, 'c': 0}

print(person)
print(person2)
print(from_keys)
```

## Accessing values

```python
print(person["name"])              # Alice
print(person.get("age"))           # 30
print(person.get("email", "N/A"))  # N/A (default if key missing)
```

## Adding / updating

```python
person["email"] = "alice@example.com"
person.update({"age": 31, "city": "LA"})
print(person)
```

## Removing items

```python
del person["city"]
removed = person.pop("email")           # removes and returns value
print("Removed:", removed)
print(person)

key, val = person.popitem()             # removes last inserted item
print("Popped item:", key, val)
print(person)

person.clear()                          # empties the dict
print(person)
```

## Iterating

```python
person = {"name": "Alice", "age": 30, "city": "New York"}

for key in person:
    print(key, person[key])

print("---")

for key, value in person.items():
    print(f"{key}: {value}")

print("---")

for value in person.values():
    print(value)
```

## Checking existence and length

```python
if "name" in person:
    print("Key exists")

print(len(person))
```

## Merging dictionaries (Python 3.9+)

```python
d1 = {"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}

merged = d1 | d2        # {'a': 1, 'b': 3, 'c': 4}
print(merged)

d1 |= d2                # in-place merge
print(d1)
```

## Dictionary comprehension

```python
squares = {x: x**2 for x in range(5)}
print(squares)   # {0:0, 1:1, 2:4, 3:9, 4:16}
```

## Nested dictionaries

```python
nested = {
    "user1": {"name": "Alice", "age": 30},
    "user2": {"name": "Bob", "age": 25}
}
print(nested["user1"]["name"])   # Alice
```

## Sorting by value

```python
scores = {"a": 3, "b": 1, "c": 2}
sorted_scores = dict(sorted(scores.items(), key=lambda x: x[1]))
print(sorted_scores)
```

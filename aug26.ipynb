```python
lllll = [[2, 5, 6], [4], [3, 4], [3, 5, 6, 7]]

x = []  # 4
y = []  # 1
for ele in lllll:
    # print(len(ele))
    lengthOfList = len(ele)
    if lengthOfList == 4:
        for ee in ele:
            x.append(ele)
    if lengthOfList == 1:
        for ee in ele:
            y.append(ele)

print("xlist >>", x)
print("ylist >>", y)
```

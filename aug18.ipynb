## special operators
### in -> membership operators (Boolean)
### is -> Identity Operator (Boolean)

```python
s = "pune"
print("e" in s)
print("z" in "pune")
```

```python
s = "programming" 
# count of characters or frequency of characters
print(s.count("g"))
```

```python
s = "programming" #count of characters or frequency of characters

count = 0
for char in s :
    if char == "g":
        count+=1
print("frequency of char g = " , count)
```

### Make below code dynamic
### Ask on console using input function which character frequency you want
## and display freq of that char

```python
s = input("Enter a String : ") 

print("Entered String:",s)
char = input("Enter char : ")
count = 0
for ch in s :
    if ch == char :
        count+=1
print("frequency of char",char, "=", count)
```

```python
s = input("Enter String:")
print("Entered String" , s)
count=0
for ch in s :
    if ch == " ":
        count+=1
print("Total white spaces : " ,count)

```

```python
original_s = input("Enter a string: ")
rev_s = original_s[::-1]

if original_s.lower() == rev_s.lower():
    print(original_s, "is a palindrome")
else:
    print(original_s, "is NOT a palindrome")
```

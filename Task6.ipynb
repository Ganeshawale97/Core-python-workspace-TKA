## Topic 1: Operators
### Relational / Comparison Operators: >, <, >=, <=
### Equality Operators: ==, !=
### Comparison operators return Boolean values (True / False).

```python
#Relational Operators

a = 10
b = 5

print(a > b)    # True
print(a < b)    # False
print(a >= b)   # True
print(a <= b)   # False

#Equality Operators

a = 10
b = 10

print(a == b)   # True
print(a != b)   # False

#Boolean Operators

jay = 21
viru = 22

print(jay > viru)   # False
print(jay != viru)  # True
print(jay == viru)  # False

# Find the largest number

a = 25
b = 40
c = 30

print(a > b)       # False
print(b > c)       # True
print(b >= a)      # True
print(c != a)      # True# 

# Compare student marks

ganesh = 85
rahul = 78

print(ganesh > rahul)     # True
print(ganesh == rahul)    # False
print(ganesh != rahul)    # True
print(ganesh >= 80)       # True


```

## Topic 2: Conditional Statements

### if statement
### if-else statement
### if-elif-else statement
### How Python checks a condition
### What happens when the if condition is False

```python
# Conditional Statements


# if statement

age = 20

if age >= 18:
    print("You can vote")


# if-else statement

age = 16

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote")


# if-elif-else statement

jay = 42
viru = 42

if jay > viru:
    print("jay will pay the bill")
elif jay == viru:
    print("both will pay the bill")
else:
    print("viru will pay the bill")


# Check grade

marks = 85

if marks >= 90:
    print("Grade A+")
elif marks >= 75:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")


# Check temperature

temperature = 35

if temperature >= 40:
    print("Very hot")
elif temperature >= 30:
    print("Hot")
elif temperature >= 20:
    print("Normal")
else:
    print("Cold")
```

## Topic 3: Logical Operators

### AND
### OR
### NOT

```python
# Logical Operators


# and operator

var1 = True
var2 = True

print(var1 and var2)    # True


var3 = False
var4 = True

print(var3 and var4)    # False


var5 = False
var6 = False

print(var5 and var6)    # False

# Logical Operators

username = "ganesh"
password = "1234"

if username == "ganesh" and password == "1234":
    print("Login successful")
else:
    print("Invalid login")


# Logical Operators

day = "Sunday"
holiday = False

if day == "Sunday" or holiday == True:
    print("No college today")
else:
    print("College today")
```

## Topic 4: or Operator

```python
# or operator

c1 = "samsung"
c2 = "apple"
c3 = "oneplus"

if (c1 == "samsung") or (c2 == "apple"):
    print("I will buy the phone")


# or operator

food1 = "pizza"
food2 = "burger"

if food1 == "pizza" or food2 == "pizza":
    print("Pizza is available")
else:
    print("Pizza is not available")
```

## Topic 5: not Operator

```python
# not operator

VAR = True

print(not VAR)    # False

# Example :

is_empty = False

if not is_empty:
    print("The box contains something")
else:
    print("The box is empty")
```

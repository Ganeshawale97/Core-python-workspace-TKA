### 25 Python Basics practical questions covering data types, operators, built-in functions, conditions, and strings.

## Questions

#### 1. Basic Variables – Create variables for name, age, percentage, and placement status and display their values and data types.

```python
name = "Rahul"
age = 22
percentage = 78.5
placed = True

print(name, type(name))
print(age, type(age))
print(percentage, type(percentage))
print(placed, type(placed))
```

#### 2. Type Conversion – Take age as input and display the age after 5 years.

```python
age = int(input("Enter age: "))

new_age = age + 5

print("Age after five years : " , new_age)
```

#### 3. Built-in Functions – Given a list of numbers, find the minimum, maximum, and total using built-in functions.

```python
num = [25, 40, 28, 67, 32]

print("minimum:", min(num))
print("Maximum:", max(num))
print("Total:", sum(num))
```

#### 4. Numeric Data Types – Demonstrate int, float, and complex data types.

```python
a = 10
b = 25.5
c = 3 + 4j

print(a, type(a))
print(b, type(b))
print(c, type(c))
```

#### 5. Length and Absolute Value – Find the length of a name and the absolute value of a negative number.

```python
name = "Ganesh"

print(len(name))
print("Absolute value:", abs(-125))
```

#### 6. Round Function – Round 78.4567 to 2 decimal places.

```python
number = 78.4567

print(round(number, 2))
print(round(number, 3))
print(round(number, 1))
print(round(number, 0))
print(int(number))

# int() removes the decimal part.
# round() rounds the number.
```

#### 7. Power Function – Take a number as input and find its square and cube using pow().

```python
number = int(input("Enter number: "))

print("Entered Number is :", number)
print("Square:", pow(number, 2))
print("Cube:", pow(number, 3))
```

#### 8. Arithmetic Operators : Take two numbers as input and perform addition, subtraction, multiplication, division, floor division, modulus, and power.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
```

#### 9. Comparison Operators Take two numbers as input and compare them using >, <, ==, !=, >=, and <=.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

#### 10. Logical Operators : Take age and percentage as input and check whether the person is at least 18 years old and has a percentage of at least 60%.

```python
age = int(input("Enter age: "))
percentage = float(input("Enter percentage: "))

result = age >= 18 and percentage >= 60

print(result)
```

#### 11. Assignment Operators : Demonstrate the assignment operators +=, -=, *=, and /= using a variable.

```python
x = 100

x += 20
print(x)

x -= 10
print(x)

x *= 2
print(x)

x /= 5
print(x)
```

#### 12. Simple Eligibility Check : Take marks as input and check whether the student is Eligible or Not Eligible.

```python
marks = float(input("Enter marks: "))

if marks >= 60:
    print("Eligible")
else:
    print("Not Eligible")
```

#### 13. Even or Odd : Take a number as input and check whether the number is Even or Odd.

```python
number = int(input("Enter number: "))

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

#### 14. String Indexing : Using the string "KiranAcademy", print the first character, last character, fifth character, and length of the string.

```python
text = "KiranAcademy"

print("First:", text[0])
print("Last:", text[-1])
print("Fifth:", text[4])
print("Length:", len(text))
```

#### 15. String Slicing : Using the string "Programming", perform string slicing to print the first 6 characters, last 4 characters, and the reversed string. 

```python
text = "Programming"

print(text[0:6])
print(text[-4:])
print(text[::-1])
```

#### 16. String Methods : Use the string " python full stack " and demonstrate the strip(), upper(), lower(), title(), and replace() methods.

```python
text = " python full stack "

print(text.strip())
print(text.upper())
print(text.lower())
print(text.title())
print(text.replace("python", "java"))
```

#### 17. String Search : Check whether "Python" is present in the string and find the position of "Testing" using find()


```python
text = "Java Python Testing Data"

print("Python" in text)
print(text.find("Testing"))
```

#### 18. String Count : Count the occurrences of "a" and "n" in the string "banana".

```python
text = "banana"

print("a count:", text.count("a"))
print("n count:", text.count("n"))
```

#### 19. startswith() and endswith() : Check whether the string "Kiran Academy Pune" starts with "Kiran" and ends with "Pune".

```python
text = "Kiran Academy Pune"

print(text.startswith("Kiran"))
print(text.endswith("Pune"))
```

#### 20. Split a String : Split the string "Java,Python,React,SQL" using the comma , . 

```python
text = "Java,Python,React,SQL"

values = text.split(",")

print(values)
```

#### 21. Join Strings : Join the strings "Python", "Java", and "SQL" using " - " as the separator.

```python
technologies = ["Python", "Java", "SQL"]

result = " - ".join(technologies)

print(result)
```

#### 22. Replace Word : Replace the word "Java" with "Python" in the string "I am learning Java".

```python
text = "I am learning Java"

new_text = text.replace("Java", "Python")

print(new_text)
```

#### 23. Case Checking : Take a word as input and check whether it is uppercase, lowercase, or neither.

```python
word = input("Enter a word: ")

if word.isupper():
    print("Uppercase")
elif word.islower():
    print("Lowercase")
else:
    print("Neither")
```

#### 24. Student Message : Take the student name, course, and marks as input. Check eligibility based on marks and display the student details with their status

```python
name = input("Enter student name: ")
course = input("Enter course: ")
marks = float(input("Enter marks: "))

if marks >= 60:
    status = "Eligible"
else:
    status = "Not Eligible"

print(f"Student: {name}, Course: {course}, Marks: {marks}, Status: {status}")
```

####  25. Simple Bill Calculation : Take the product name, quantity, and price per unit as input and calculate the total bill.

```python
product = input("Enter product name: ")
quantity = int(input("Enter quantity: "))
price = float(input("Enter price per unit: "))

total = quantity * price

print(f"Product: {product}")
print(f"Quantity: {quantity}")
print(f"Price per Unit: {price}")
print(f"Total Bill: {total}")
```

#### A set is an unordered, mutable collection of unique, hashable items.

## Q. How to access elements of set data type?
### -> by using for loop

```python
for data in s:
    print(data)
```

```python
s = set()

s.add("Ganesh")
s.add(234)
s.add(34)
s.add((5, 6)) #tuple
s.add(tuple(["akash", "preeti", "pranav"])) #list


print(s)
```

```python
l = [34,56,45,73,24,75,91,23,45,91]

print(max(l))

l.sort()
print(l)

print(l[-1]) #last num
print(l[-2]) #second last num
```

```python
l = [34,56,45,73,24,75,91,23,45,91]

s1 = set(l)
print(s1)
l1 = list(s1)
l1.sort()
print(l1)

print("Max Number:", l1[-1])
print("Second Max Number:", l1[-2])

print("Min Number:", l1[0])
print("Second Min Number:", l1[1])
```

```python
s1 ={1,2,3,4,5}
s2 ={4,5,6,7,8}

s3 = s1.intersection(s2)
s4 = s1.union(s2)

print(s1.intersection(s2))
print(s1.union(s2))

print(s3)
print(s4)

print(s1.pop())
print(s1.remove(3))
```

## Q. What is frozen set?
### -> set is mutable but frozen set is im-mutable.

```python
s={2,3,4,5,6,7}
fs = frozenset(s)
print(fs)
print(type(fs))
print(dir(fs))
```

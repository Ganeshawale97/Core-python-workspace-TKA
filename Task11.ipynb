# Explore 5 List Methods
### Homework

## Setup
### Continuing with our batch list from class

```python
batch_1341_db = ["Surbhi", "kunal", "Ganesh", "Srushti", "Supriya", "Soniya", "ketak", "Prerena", "Sanika"]
print(batch_1341_db)
```

## 1. insert()
### Adds an element at a specific index (does NOT overwrite, shifts other elements right)
### Syntax: list.insert(index, element)

```python
batch_1341_db.insert(2, "Rohit")   # insert 'Rohit' at index 2
print(batch_1341_db)
print(len(batch_1341_db))
```

## 2. extend()
### Adds ALL elements of another iterable (list/tuple/string) to the end, one by one
### Difference from append(): append() adds the whole thing as ONE element, extend() unpacks it
### Syntax: list.extend(iterable)

```python
new_students = ["Aditi", "Mahesh"]

batch_1341_db.append(new_students)   # appends as a single nested list
print("After append:", batch_1341_db)

batch_1341_db.remove(new_students)   # undo, so we can compare cleanly

batch_1341_db.extend(new_students)   # extends -> each name added individually
print("After extend:", batch_1341_db)
```

## 3. remove()
### Removes the FIRST matching element by value (not by index)
### Raises ValueError if the element is not found
### Syntax: list.remove(value)

```python
print("Before remove:", batch_1341_db)
batch_1341_db.remove("ketak")   # removes by value
print("After remove:", batch_1341_db)
```

## 4. pop()
### Removes AND returns an element at a given index (default: last element)
### Useful when you need the removed value for further use
### Syntax: list.pop(index=-1)

```python
last_student = batch_1341_db.pop()        # removes & returns last element
print("Popped (last):", last_student)

first_student = batch_1341_db.pop(0)      # removes & returns element at index 0
print("Popped (index 0):", first_student)

print("List after pops:", batch_1341_db)
```

## 5. sort()
### Sorts the list IN-PLACE (modifies original list, returns None)
### Default: ascending order. Use reverse=True for descending
### Note: sort() only works if all elements are comparable (e.g. all strings, or all numbers)
### Syntax: list.sort(reverse=False, key=None)

```python
names_only = []
for item in batch_1341_db:
    if isinstance(item, str):
        names_only.append(item)

print("Before sort:", names_only)

names_only.sort()                # ascending (alphabetical)
print("Ascending:", names_only)

names_only.sort(reverse=True)    # descending
print("Descending:", names_only)
```

## Summary
| Method | Purpose | Modifies list? | Returns |
|---|---|---|---|
| insert(i, x) | add x at index i | Yes | None |
| extend(iterable) | add all elements of iterable | Yes | None |
| remove(x) | remove first occurrence of value x | Yes | None |
| pop(i) | remove & return element at index i | Yes | removed element |
| sort() | sort list in-place | Yes | None |

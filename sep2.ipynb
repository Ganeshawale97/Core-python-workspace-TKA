```python
db = {1: 'Jay', 2: 'Pavan', 3: 'Raj', 4: 'Rina'}

print("1)Using .startswith()")
for roll in db:
    if db[roll].startswith('R'):
        print(f"Roll No: {roll}, Name: {db[roll]}")

print("-------------")

print("2)Using name[0]")
for roll in db:
    if db[roll][0] == 'R':
        print(f"Roll No: {roll}, Name: {db[roll]}")

print("-------------")

print("3)Using .items()")
for roll, name in db.items():
    if name.startswith('R'):
        print(f"Roll No: {roll}, Name: {name}")
```

```python
db = {1: 'Jay', 2: 'Pavan', 3: 'Raj', 4: 'Rina'}

print("1) Using 'in' operator")
for roll in db:
    if 'n' in db[roll]:
        print(f"Roll No: {roll}, Name: {db[roll]}")

print("-------------")

print("2) Using .items()")
for roll, name in db.items():
    if 'n' in name:
        print(f"Roll No: {roll}, Name: {name}")

print("-------------")

print("3) Using db.get()")
for roll in db:
    if 'n' in db.get(roll):
        print(f"Roll No: {roll}, Name: {db[roll]}")

print("-------------")

print("4) Using .find()")
for roll in db:
    if db[roll].find('n') != -1:
        print(f"Roll No: {roll}, Name: {db[roll]}")

print("-------------")

print("5) Using __contains__()")
for roll in db:
    if db[roll].__contains__('n'):
        print(f"Roll No: {roll}, Name: {db[roll]}")
```

```python
movies2026_db = {
    'Aavarapan 2': ["Emraan Hashmi", "Disha Patani"],
    'Pathaan 2': ["Shah Rukh Khan", "Deepika Padukone"],
    'War 2': ["Hrithik Roshan", "Jr NTR"],
    'Singham Again': ["Ajay Devgn", "Kareena Kapoor"],
    'Don 3': ["Ranveer Singh", "Kiara Advani"],
    'Coolie': ["Rajinikanth", "Nagarjuna"],
    'Kalki 2898 AD Part 2': ["Prabhas", "Deepika Padukone"],
    'Baaghi 4': ["Tiger Shroff", "Sanjay Dutt"],
    'Alpha': ["Alia Bhatt", "Sharvari"],
    'Metro In Dino': ["Aditya Roy Kapur", "Sara Ali Khan"],
}

print("1) Display all movies with cast")
for movie, cast in movies2026_db.items():
    print(f"{movie}: {', '.join(cast)}")

print("-------------")

print("2) Display only movie titles")
for movie in movies2026_db:
    print(movie)

print("-------------")

print("3) Display only movie titles using .keys()")
for movie in movies2026_db.keys():
    print(movie)

print("-------------")

print("4) Movies where a specific actor appears")
actor = "Deepika Padukone"
for movie, cast in movies2026_db.items():
    if actor in cast:
        print(movie)

print("-------------")

print("5) Movies with exactly 2 lead actors")
for movie, cast in movies2026_db.items():
    if len(cast) == 2:
        print(movie)

print("-------------")

print("6) Total number of movies")
print(f"Total movies: {len(movies2026_db)}")

print("-------------")

print("7) All unique actors across all movies")
all_actors = set()
for cast in movies2026_db.values():
    all_actors.update(cast)
print(all_actors)

print("-------------")

print("8) Movies whose title contains a number (sequels)")
for movie in movies2026_db:
    if any(char.isdigit() for char in movie):
        print(movie)

print("-------------")

print("9) Movies sorted alphabetically")
for movie in sorted(movies2026_db):
    print(movie)

print("-------------")

print("10) Display actors of Don 3")
for actor in movies2026_db['Don 3']:
    print(actor)

print("-------------")

print("11) Display actors of Don 3 (nested loop version)")
for movie in movies2026_db:
    if movie == "Don 3":
        for nm in movies2026_db[movie]:
            print(nm)

print("-------------")

print("12) Total number of 'a' in Coolie")
movie = "Coolie"
count = movie.lower().count('a')
print(f"Total 'a' in {movie}: {count}")

print("-------------")

print("13) Total number of 'a' in all movie names")
total_a = 0
for movie in movies2026_db:
    total_a += movie.lower().count('a')
print(f"Total 'a' in all movie names: {total_a}")
```

## Dict inside Dict

```python
movies2026_db = {
    'Aavarapan 2': {'cast': ["Emraan Hashmi", "Disha Patani"], 'year': 2026, 'genre': 'Thriller'},
    'Pathaan 2': {'cast': ["Shah Rukh Khan", "Deepika Padukone"], 'year': 2026, 'genre': 'Action'},
    'War 2': {'cast': ["Hrithik Roshan", "Jr NTR"], 'year': 2026, 'genre': 'Action'},
    'Singham Again': {'cast': ["Ajay Devgn", "Kareena Kapoor"], 'year': 2026, 'genre': 'Action'},
    'Don 3': {'cast': ["Ranveer Singh", "Kiara Advani"], 'year': 2026, 'genre': 'Thriller'},
    'Coolie': {'cast': ["Rajinikanth", "Nagarjuna"], 'year': 2026, 'genre': 'Action'},
    'Kalki 2898 AD Part 2': {'cast': ["Prabhas", "Deepika Padukone"], 'year': 2026, 'genre': 'Sci-Fi'},
    'Baaghi 4': {'cast': ["Tiger Shroff", "Sanjay Dutt"], 'year': 2026, 'genre': 'Action'},
    'Alpha': {'cast': ["Alia Bhatt", "Sharvari"], 'year': 2026, 'genre': 'Action'},
    'Metro In Dino': {'cast': ["Aditya Roy Kapur", "Sara Ali Khan"], 'year': 2026, 'genre': 'Drama'},
}

print("1) Display all movies with year and genre")
for movie, details in movies2026_db.items():
    print(f"{movie} ({details['year']}) - {details['genre']}")

print("-------------")

print("2) Display all movies with cast, year, genre")
for movie, details in movies2026_db.items():
    print(f"{movie} ({details['year']}) - {details['genre']}")
    print(f"  Cast: {', '.join(details['cast'])}")

print("-------------")

print("3) Cast of a specific movie (Don 3)")
print(movies2026_db['Don 3']['cast'])

print("-------------")

print("4) Year of a specific movie (Don 3)")
print(movies2026_db['Don 3']['year'])

print("-------------")

print("5) Genre of a specific movie (Don 3)")
print(movies2026_db['Don 3']['genre'])

print("-------------")

print("6) All Action movies")
for movie, details in movies2026_db.items():
    if details['genre'] == 'Action':
        print(movie)

print("-------------")

print("7) All Thriller movies")
for movie, details in movies2026_db.items():
    if details['genre'] == 'Thriller':
        print(movie)

print("-------------")

print("8) Movies released in 2026")
for movie, details in movies2026_db.items():
    if details['year'] == 2026:
        print(movie)

print("-------------")

print("9) Movies where a specific actor appears")
actor = "Deepika Padukone"
for movie, details in movies2026_db.items():
    if actor in details['cast']:
        print(movie)

print("-------------")

print("10) Count of movies per genre")
genre_count = {}
for details in movies2026_db.values():
    genre = details['genre']
    genre_count[genre] = genre_count.get(genre, 0) + 1
print(genre_count)

print("-------------")

print("11) Add a new movie with full details")
movies2026_db['Dhurandhar 2'] = {'cast': ["Ranveer Singh", "Sara Arjun"], 'year': 2026, 'genre': 'Action'}
print(movies2026_db['Dhurandhar 2'])

print("-------------")

print("12) Add an actor to an existing movie's cast")
movies2026_db['Coolie']['cast'].append("Amitabh Bachchan")
print(movies2026_db['Coolie']['cast'])

print("-------------")

print("13) List of all genres (unique)")
genres = set()
for details in movies2026_db.values():
    genres.add(details['genre'])
print(genres)

print("-------------")

print("14) List of all movie titles sorted by year")
for movie in sorted(movies2026_db, key=lambda m: movies2026_db[m]['year']):
    print(movie, movies2026_db[movie]['year'])

print("-------------")

print("15) List of all movie titles sorted alphabetically")
for movie in sorted(movies2026_db):
    print(movie)

print("-------------")

print("16) Total number of actors across all movies")
total_actors = 0
for details in movies2026_db.values():
    total_actors += len(details['cast'])
print(f"Total actors: {total_actors}")

print("-------------")

print("17) Movies with more than 2 actors")
for movie, details in movies2026_db.items():
    if len(details['cast']) > 2:
        print(movie)

print("-------------")

print("18) Check if a movie exists in the db")
movie_name = "War 2"
if movie_name in movies2026_db:
    print(f"{movie_name} found!")
else:
    print(f"{movie_name} not found.")

print("-------------")

print("19) Check if an actor exists anywhere in the db")
actor_name = "Alia Bhatt"
found = False
for details in movies2026_db.values():
    if actor_name in details['cast']:
        found = True
        break
print(f"{actor_name} found: {found}")

print("-------------")

print("20) Update the genre of a movie")
movies2026_db['Alpha']['genre'] = 'Sci-Fi'
print(movies2026_db['Alpha'])

print("-------------")

print("21) Remove a movie from the db")
movies2026_db.pop('Metro In Dino')
print(list(movies2026_db.keys()))

print("-------------")

print("22) Movies grouped by genre")
grouped = {}
for movie, details in movies2026_db.items():
    genre = details['genre']
    grouped.setdefault(genre, []).append(movie)
print(grouped)

print("-------------")

print("23) Longest movie title")
longest = max(movies2026_db, key=len)
print(longest)

print("-------------")

print("24) All movie titles that are sequels (contain a digit)")
for movie in movies2026_db:
    if any(char.isdigit() for char in movie):
        print(movie)

print("-------------")

print("25) All actors sorted alphabetically (no duplicates)")
all_actors = set()
for details in movies2026_db.values():
    all_actors.update(details['cast'])
print(sorted(all_actors))
```

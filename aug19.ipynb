## What is list?
### List is a mutable, heterogeneous collection of elements where order is maintained and duplicates are allowed.
### values enclosed in square bracket[]


### Mutuable ----> changable, growable
### Heterogenious Collection of elements --> int, float, string, boolean, complex, list, tuple , dict
### Order is maintained ------------> Index is present (indexing & slicing & step value, + & - index)
### Duplicates are allowed

## Q. How to access data from list?
### By using indexing & Silicing

```python
print(batch_1341_db[0]) #indexing
print(batch_1341_db[-1]) #indexing
print(batch_1341_db[2:5]) #Slicing
print(batch_1341_db[::-1]) #reverse
print(batch_1341_db[::-2]) #reverse_step
```

## Q.How to iterate every element form list?
### by using loop -> for, while

```python
for name in batch_1341_db:
    if type(name) == str:
        print(name.upper())
        
```

```python
filtered_list = []

for name in batch_1341_db:
    if isinstance(name, str):
        filtered_list.append(name)

print(filtered_list)
```

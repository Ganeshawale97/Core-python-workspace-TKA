# Advanced String Methods Explorer

### STRING INFORMATION

```python
text = input("Enter a string: ")

print("\n========== STRING INFORMATION ==========")

print("Original String       :", text)
print("Length                :", len(text))
print("First Character       :", text[0] if text else "Empty")
print("Last Character        :", text[-1] if text else "Empty")
```

### CASE METHODS

```python
print("\n========== CASE METHODS ==========")

print("Upper                 :", text.upper())
print("Lower                 :", text.lower())
print("Title                 :", text.title())
print("Capitalize            :", text.capitalize())
print("Swapcase              :", text.swapcase())
print("Casefold              :", text.casefold())
```

### CHECKING METHODS

```python
print("\n========== CHECKING METHODS ==========")

print("Is Alpha              :", text.isalpha())
print("Is Digit              :", text.isdigit())
print("Is Alphanumeric       :", text.isalnum())
print("Is Decimal            :", text.isdecimal())
print("Is Lower              :", text.islower())
print("Is Upper              :", text.isupper())
print("Is Space              :", text.isspace())
print("Is Title              :", text.istitle())
```

### SEARCH METHODS

```python
print("\n========== SEARCH METHODS ==========")

print("Count of 'a'          :", text.lower().count("a"))
print("Find 'a'              :", text.lower().find("a")) #find() → returns -1 if not found.
print("Index of 'a'          :", text.lower().find("a")) #index() → raises ValueError if not found.
```

### START / END

```python

print("\n========== START / END ==========")

print("Starts with 'A'       :", text.startswith("A"))
print("Ends with 'z'         :", text.endswith("z"))
```

### MODIFYING METHODS

```python
print("\n========== MODIFYING METHODS ==========")

print("Replace spaces        :", text.replace(" ", "_")) #string.replace(old, new)
print("Replace 'a' with '@'  :", text.replace("a", "@")) #string.replace(old, new, count)
print("Strip                 :", text.strip())
print("Left Strip            :", text.lstrip())
print("Right Strip           :", text.rstrip())
```

### SPLIT / JOIN

```python
print("\n========== SPLIT / JOIN ==========")

words = text.split()

print("Split into words      :", words)
print("Number of words       :", len(words))
print("Joined with '-'       :", "-".join(words))
```

### ALIGNMENT

```python

print("\n========== ALIGNMENT ==========")

print("Left aligned          :", text.ljust(20, "-"))
print("Right aligned         :", text.rjust(20, "-"))
print("Center aligned        :", text.center(20, "-"))
```

### PARTITION

```python
print("\n========== PARTITION ==========")

print("Partition by space    :", text.partition(" "))
print("RPartition by space   :", text.rpartition(" "))
```

### STRING SLICING

```python
print("\n========== STRING SLICING ==========")

print("First 3 characters    :", text[:3])
print("Last 3 characters     :", text[-3:])
print("Reverse               :", text[::-1])
print("Every 2nd character   :", text[::2])
```

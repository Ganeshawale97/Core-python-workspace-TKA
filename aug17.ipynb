## String slicing with step value/size
### [start : end : step]
### step value is optional
### If we dont write step calue then by default it is +1
    -->if step value sign is posiitive then last index -1
    -->if step valur sign is negative then last index +1

```python
s = "instagram"
print(s[1:8]) #nstagra
print(s[1:8:2]) #ntga  
print(s[8 : -1 : -1]) #last index + 1
print(s[0 : 9 : +1]) #last index - 1
print(s[0 : : 1])
print(s[ : : 1])
```

```python
email = "rahul.sharma@gmail.com"
username = email[ :  12 : +1]
domain = email[ 13 :  : +1]
print("Username = " , username)
print("Domain = " , domain)
```

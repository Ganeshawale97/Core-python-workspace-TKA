## Q. What is dict in Python ?
### → Dict are Mutable , Unordered (before python 3.7), ordered (after python 3.7) , Heterogeneous collection of elements where data is stored in key-value pair
### Unordered --> Index is not present
### Keys must be immutable (5 fundamental, tuple, fs) so they can be hashed

### Dict is very imp data type -> JSON data type is widely used in web applications
### Data bases -> NoSQL -> No tables -> MongoDB -> Pair -> Key value

## Dictonary methods
### 1. keys() --> It will return all dict keys in list
### 2. Values() --> It will return all dict Values in list
### 3. items() --> It will return all dict keys and values in tuple of list

```python
print(dict.keys())
for i in dict.keys():
    print(i)
```

```python
dict[1] = "Hello"
dict[1.5] = "Hello"
dict[(1,2,3,4,5)] = "Hello"
dict[True] = "Hello"
print(dict)
print(dict["city"])
```

```python
dict = {1:"jay", 2:"kumar", 3:"sai", 4:"ram", 5:"krishna"}
print(dict)
print(dict[1])
print(dict.get(2))
for k in dict:
    print(k, "---->", dict.get(k))

```

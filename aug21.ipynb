# Tuple

## What is Tuple?
### -> Tuple is a im-mutable , heterogenious collection of elements where order is maintained and duplicates are allowed

#### add , delete , pop , remove , insert , append --> not available in tuple

# Q. How to create Tuple?
### sty:-> t = (1,2,3,4,5,6)

```python
t = (10,20,30,40)
print(dir(t))
print(t.__str__())
print(t.count(10))
print(t.index(30))
```

## Create a tuple with single element?

```python
x = (10,)
print(x.__str__())
print(x.count(10))
print(x.index(10))
print(x.__class__)
```

# Q. How to access data from tuple?
## ->Indexing & Silicing 

```python
category = ("electronics" , "furniture" , "electronics" , "clothing")
print(category)

# change/ update -> tuple into list --> do changes --> list into tuple

cat_list = list(category)
print(cat_list)
cat_list.append("stationary")
print(cat_list)
category = tuple(cat_list)
print(category)


print(id(category))
print(id(category[0]))
print(id(category[1]))
print(id(category[2]))
```

# Packing & Unpacking Of elements / Data

```python
bag1 = ("clothes" , "charger" , "Cash", "whatever") #tuple
bag2 = ["clothes" , "charger" , "Cash", "whatever"] #list

print(bag1)
v1, v2, v3, v4 = bag1 #unpacking
print(v1, v2, v3, v4)
```

```python
t1 = "a"
t2 = 20
t3 = "b"
t4 = "z"

bag1 = t1, t2, t3, t4 #packing of elements
print(bag1)
print(type(bag1))
```

```python
products = ["mobile" , "charger" , "Notebook" , "T-shirt"]
category = ("electronics", "electronics" , "stationary", "clothing")

c1, c2, c3, c4 = category
p1, p2, p3, p4 = products

print(c1, "--->", p1)
print(c2, "--->", p2)
print(c3, "--->", p3)
print(c4, "--->", p4)
```

```python
products = ["mobile" , "charger" , "Notebook" , "T-shirt"]
category = ("electronics", "electronics" , "stationary", "clothing")

for i,p in enumerate(products):
    print(category[i], "--->", p)
```

```python
count = 0

for p in products:
    print(category[count], "---->", p)
    count += 1
```

### If you want to display only the product name for the "Stationary" category:


```python
products = ["Mobile", "Charger", "Notebook", "T-shirt"]
category = ("Electronics", "Electronics", "Stationary", "Clothing")

count = 0

for p in products:
    if category[count] == "Stationary":
        print(p)
    count += 1
```

# Q. What is Range?
### -> Range is data type in python.
### -> It is used to create sequence of integer numbers

## we can create a range by following ways 
### 1. range(N) 
### 2. range(start, end)
### 3. range(start, end , step)

```python
# Q. How to create Range?

N = 12
r = range(N)
print(r)
print(type(r))
```

```python
for i in r:
    print(i)
```

```python
r = range(50 , 67)
for num in r:
    print(num)
```

```python
for num in range(5, 50 , 5):
    print(num)
```

```python
for i in range(-1 , -10, -1):
    print(i)
```

# write a program to print 100 to 1 using range function

```python
for i in range(100, 0, -1):
    print(i)
```

```python
a = int(input("Enter start:"))
print("Start :" , )
b = int(input("Enter end:"))
c = int(input("Enter step:"))


for num in range(a, b, c):
    print(num)
```

## WAP to print all even numbers from 112 to 122

```python
for i in range(112, 123, 2):
    print(i)

print("------")

for i in range(112, 123):
    if i % 2 == 0:
        print(i)

print("------")

for i in range(112, 123):
    if i % 2 != 1:
        print(i)
```

## String ,  list, tuple...

```python
s = "instagram"

for t in enumerate(s):
     print(t)
     i , ch = t
     print(i , "------>" , ch)

print("================")

for i in range ( 0, len(s)//2 ):
     print(i, "-->" , ch)
```

```python
s = "instagram"


for i in range(len(s) - 1, - 1, -1):
    print(i, "-->", s[i])
```

```python
s = "instagram"


for i in range(0 , len(s) ):
    print(i, "-->", s[len(s)-1-i])
```

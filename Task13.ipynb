# Lists & Tuples: Problems with Solutions

### Problem 1
Create two lists `l1 = [10, 20, 30]` and `l2 = [40, 50, 60]`. Combine them into one list using `+` and print the result.

````python
l1 = [10, 20, 30]
l2 = [40, 50, 60]
print(l1 + l2)
````

### Problem 2
Create a list `stock = [12, 45, 67, 23, 89]`. Remove the value `45` using `.remove()`, then remove the last element using `.pop()`. Print the list after each step.

````python
stock = [12, 45, 67, 23, 89]
stock.remove(45)
print(stock)
stock.pop()
print(stock)
````

### Problem 3
Create a tuple `colors = ("red", "green", "blue", "red", "yellow")`. Print how many times `"red"` appears, and the index of `"blue"`.

````python
colors = ("red", "green", "blue", "red", "yellow")
print(colors.count("red"))
print(colors.index("blue"))
````

### Problem 4

````python
x = ("solo",)
print(x)
print(type(x))
````

### Problem 5
Create a tuple `city = ("Pune", "Mumbai", "Delhi")`. Try adding `"Chennai"` directly (it will error), then add it correctly by converting to a list and back.

````python
city = ("Pune", "Mumbai", "Delhi")

# city.append("Chennai")  # AttributeError: tuple has no attribute 'append'

city_list = list(city)
city_list.append("Chennai")
city = tuple(city_list)
print(city)
````

### Problem 7
Pack four values `10, 20, 30, 40` into a tuple `nums` using comma-based packing (no explicit parentheses), then print `nums` and its type.

### Problem 8
Given:
```python
employees = ["Aman", "Riya", "Karan", "Simran"]
departments = ("Sales", "IT", "HR", "IT")
```
Use `enumerate()` to print each employee with their department, e.g. `Aman - Sales`.

````python
employees = ["Aman", "Riya", "Karan", "Simran"]
departments = ("Sales", "IT", "HR", "IT")

for i, emp in enumerate(employees):
    print(emp, "-", departments[i])
````

### Problem 9
Repeat Problem 8 without `enumerate()`, using a manual counter.

````python
employees = ["Aman", "Riya", "Karan", "Simran"]
departments = ("Sales", "IT", "HR", "IT")

count = 0
for emp in employees:
    print(emp, "-", departments[count])
    count += 1
````

### Problem 10
Given:
```python
products = ["Mobile", "Charger", "Notebook", "Pen", "T-shirt"]
category = ("Electronics", "Electronics", "Stationary", "Stationary", "Clothing")
```
Print only the products where category is `"Electronics"`.

````python
products = ["Mobile", "Charger", "Notebook", "Pen", "T-shirt"]
category = ("Electronics", "Electronics", "Stationary", "Stationary", "Clothing")

count = 0
for p in products:
    if category[count] == "Electronics":
        print(p)
    count += 1
````

### Problem 11
Given `scores = (55, 72, 88, 40, 91)`, remove the lowest score, append `100`, and print the final result as a tuple.

````python
scores = (55, 72, 88, 40, 91)

scores_list = list(scores)
scores_list.remove(min(scores_list))
scores_list.append(100)
scores = tuple(scores_list)
print(scores)
````

### Problem 12
Given `t = (1, 2, 3, 4, 5)`, create a new tuple containing only the even numbers.

````python
t = (1, 2, 3, 4, 5)

even_list = [n for n in t if n % 2 == 0]
even_tuple = tuple(even_list)
print(even_tuple)
````
